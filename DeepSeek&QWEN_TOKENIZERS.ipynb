# **DEEPSEEK = "deepseek-ai/DeepSeek-V3.1"**



```python
DEEPSEEK = "deepseek-ai/DeepSeek-V3.1"
```

```python
from transformers import AutoTokenizer

deepseek_tokenizer = AutoTokenizer.from_pretrained(DEEPSEEK)

text = "What is machine learning?"


deepseek_tokens = deepseek_tokenizer.encode(text)

print("Token IDs:")
print(deepseek_tokens)


deepseek_text = deepseek_tokenizer.decode(deepseek_tokens)

print("Decoded text:")
print(deepseek_text)


messages = [
    {"role": "user", "content": text}
]

deepseek_inputs = deepseek_tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    return_tensors="pt",
    add_generation_prompt=True
)

print("Chat template output:")
print(deepseek_inputs)
```

# **QWEN CODER MODEL**

```python
QWEN_CODER = "Qwen/Qwen2.5-Coder-7B-Instruct"
```

```python
from transformers import AutoTokenizer

qwen_tokenizer = AutoTokenizer.from_pretrained(QWEN_CODER)

text = "What is machine learning?"

# Encode: text → token IDs
qwen_tokens = qwen_tokenizer.encode(text)

print("Token IDs:")
print(qwen_tokens)

# Decode: token IDs → text
qwen_text = qwen_tokenizer.decode(qwen_tokens)

print("Decoded text:")
print(qwen_text)

# Chat template + tokenize
messages = [
    {"role": "user", "content": text}
]

qwen_inputs = qwen_tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    return_tensors="pt",
    add_generation_prompt=True
)

print("Chat template output:")
print(qwen_inputs)
```
